#RSA Algorithm

```python
import math

p=3
q=7

n=p*q
print("n = ", n)

phi = (p-1)*(q-1)
print("phi = ", phi)

e=2
while(e<phi):
  if(math.gcd(e,phi)==1):
    break
  else:
    e=e+1
print("e = ", e)\

k=2
d=((k*phi)+1)/e
print("d = ",d)
print(f"Public key: {e,n}")
print(f"Private key: {d,n}")

msg = 11
print(f"Original message: {msg}")

cipher = (msg**e)%n
print(f"Encrypted message: {cipher}")

decrypt = (cipher**d)%n
print(f"Decrypted message: {int(decrypt)}")
```

#Caesar Cipher

```python
def caesar_cipher(text, shift, mode='encrypt'):

    if mode == 'decrypt':
        shift = -shift

    result = ""
    for char in text:
        if 'a' <= char.lower() <= 'z':
            is_upper = char.isupper()

            offset = ord('A') if is_upper else ord('a')
            new_pos = (ord(char) - offset + shift) % 26
            result += chr(new_pos + offset)
        else:
            result += char
    return result

plaintext = "Hello, World!"
key = 3

encrypted_text = caesar_cipher(plaintext, key, mode='encrypt')
print(f"Original Text: {plaintext}")
print(f"Shift Key: {key}")
print(f"Encrypted Text: {encrypted_text}")

decrypted_text = caesar_cipher(encrypted_text, key, mode='decrypt')
print(f"Decrypted Text: {decrypted_text}")
```
