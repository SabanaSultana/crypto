#Digital Signature

```python
import hashlib
import math

def exteuclid(a, b):
    r1, r2 = a, b
    t1, t2 = 0, 1

    while r2:
        q = r1 // r2
        r1, r2 = r2, r1 - q * r2
        t1, t2 = t2, t1 - q * t2

    return r1, t1 % a



p = 61
q = 53
n = p * q
Pn = (p-1)*(q-1)

e = int(313)

if math.gcd(e, Pn) != 1:
    print(f"Error: Public exponent e ({e}) is not coprime to Pn ({Pn}). Please choose a different e.")
else:
    r, d = exteuclid(Pn, e)
    if r == 1:
        print("decryption key is: ", d)

    else:
        print("Multiplicative inverse not available")

original_message = "Hello World, this is a secret message!"

message_bytes = original_message.encode('utf-8')
hasher = hashlib.sha256()
hasher.update(message_bytes)
M_hash = int(hasher.hexdigest(), 16) % n

print(f"\nOriginal Message: {original_message}")
print(f"Hashed Message (M_hash): {M_hash}")

S = (M_hash**d) % n
print(f"Signature (S): {S}")

M1 = (S**e) % n
print(f"Decrypted Hash (M1): {M1}")


if M_hash == M1:
    print("Signature Valid")
else:
    print("Signature Invalid")
```

#Diffie Hellman

```python
def calculate_power(base, exponent, modulus):
   if exponent == 1:
      return base
   else:
      return pow(base, exponent) % modulus

prime = 13
generator = 19
private_key_1 = 26
private_key_2 = 15

x = calculate_power(generator, private_key_1, prime)
y = calculate_power(generator, private_key_2, prime)
secret_key_1 = calculate_power(y, private_key_1, prime)
secret_key_2 = calculate_power(x, private_key_2, prime)

print("Prime number (P):", prime)
print("Generator value (G):", generator)
print("1's private key (a):", private_key_1)
print("2's private key (b):", private_key_2)
print("Secret key for 1:", secret_key_1)
print("Secret key for 2:", secret_key_2)
```
